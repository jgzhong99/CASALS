# Refh classification and reference limits

This synthetic run demonstrates package class semantics only. A 3DEP transferred label is a pseudo-reference for comparison, not truth and not an independent accuracy certificate. The example below does not calculate accuracy or select a model.

```python
import numpy as np

from casals_l1b.classification import (
    CLASS_NAME_MAP, CLASS_REASON_MAP, DEFAULT_CONFIG, classify_points_baseline,
)
```

```python
z = np.array([100.2, 105.0, 99.0, 103.0])
ground_z = np.array([100.0, 100.0, 100.0, np.nan])
dtm_valid = np.array([True, True, True, False])
density = np.array([1.0, 1.0, 1.0, 1.0])
config = {
    **DEFAULT_CONFIG,
    "CLASSIFIER_MODE": "height_only",
    "GROUND_RESID_TOL_M": 1.5,
    "NOISE_HAG_MAX_M": 20.0,
}
result = classify_points_baseline(z, ground_z, dtm_valid, density, config)
for index, code in enumerate(result["pred_class_baseline"]):
    code = int(code)
    print(index, code, CLASS_NAME_MAP[code], CLASS_REASON_MAP[int(result["classification_reason"][index])])
```

The array is synthetic and only checks how the configured deterministic rules map height-above-ground to the existing class and reason codes. Scientific evaluation requires a declared comparison population, aligned inputs, and an explicit reference provenance; transferred 3DEP labels remain pseudo-reference.
