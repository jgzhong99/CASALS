# CASALS L1B Single H5 Beam Reader

Read one CASALS L1B H5 file and preview beam-related fields with minimal code.

```python
from pathlib import Path
import h5py
import numpy as np
import pandas as pd

H5_PATH = Path('../casals_h5_downloads/casals_l1b_20241112T165718_001_02.h5')
assert H5_PATH.exists(), H5_PATH
```

```python
with h5py.File(H5_PATH, 'r') as h5:
    keys = sorted(h5.keys())
    rx_shape = h5['rx_waveform'].shape
    beam_arrays = {name: h5[name][:] if name in h5 else None for name in [
        'local_beam_azimuth',
        'local_beam_elevation',
        'local_beam_azimuth_error',
        'local_beam_elevation_error',
    ]}
    rx_bins = h5['rx_bins'][:]
    rx_waveform = h5['rx_waveform'][0]

print('root keys:', keys)
print('rx_waveform shape:', rx_shape)
```

```python
beam_summary = pd.DataFrame([
    {
        'field': name,
        'shape': None if values is None else values.shape,
        'dtype': None if values is None else str(values.dtype),
        'preview': None if values is None else np.asarray(values[:5]).tolist(),
    }
    for name, values in beam_arrays.items()
])
beam_summary
```
