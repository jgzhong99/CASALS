# Waveform component diagnostics

This runnable synthetic example exercises the package peak detector. Its output is a list of waveform features. It does not infer extra geolocated returns: each pulse still has one official L1B `refh` point.

```python
import numpy as np

from casals_l1b.waveform import detect_candidate_components_1d, safe_argmax_bin
```

```python
bins = np.arange(128, dtype=np.float64)
waveform = (
    0.2
    + 22.0 * np.exp(-0.5 * ((bins - 34.0) / 2.2) ** 2)
    + 16.0 * np.exp(-0.5 * ((bins - 79.0) / 3.0) ** 2)
)
config = {
    "background_mode": "raw",
    "smoothing_method": "none",
    "min_height_sigma": 3.0,
    "min_prominence_sigma": 2.0,
    "min_width_bins": 1.0,
    "min_distance_bins": 5,
}
components = detect_candidate_components_1d(
    waveform, bg_mean=None, bg_std=1.0, refh_thres=None, config=config
)
main_bin, main_amplitude = safe_argmax_bin(waveform)
print({"argmax_bin": main_bin, "argmax_amplitude": main_amplitude})
print([
    {"peak_bin": item["peak_bin"], "amplitude": item["amplitude_processed"]}
    for item in components
])
```

The detector parameters above are an explicit demonstration configuration. They are not a change to the project defaults or a recommended production threshold. Real-data comparisons must preserve the existing detector baseline and its known near-tie behavior.
