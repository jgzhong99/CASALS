# Beam angle conventions are hypotheses

The H5 beam-angle encoding and the bin-to-distance interpretation are still research questions. This notebook applies one explicit candidate convention to a synthetic angle. Its output is not a selected physical convention, a refh correction, or an externally validated geolocation.

```python
import numpy as np

from research.geolocation.geolocate_sweep_bins_from_beam_angle_rules import (
    AngleRule, direction_vectors_from_rule,
)
```

```python
candidate = AngleRule(
    unit="degree",
    az_zero="north",
    az_rotation="clockwise",
    el_convention="elevation_from_horizon",
    sign="as_encoded",
    basis="instrument",
    angle_transform="as_named",
)

# Synthetic 0-degree north azimuth and 45-degree elevation.
enu_direction = direction_vectors_from_rule(
    np.array([0.0]), np.array([45.0]), candidate
)[0]
print("candidate ENU unit vector:", enu_direction)
print("unit length:", np.linalg.norm(enu_direction))
```

This convention is only a reproducible example of one candidate. The research script retains alternative units, axes, signs, angle transforms, and ray bases. Internal closure among CASALS fields must remain distinct from independent external accuracy.
